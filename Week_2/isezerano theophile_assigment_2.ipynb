# Week 2 — How Computers "Read"
### Chatbots Development (AITCD001) · Learning Outcome 1

Last week your bot matched exact keywords. Today we look at the first real step towards something smarter:
how a computer breaks a sentence into pieces it can actually work with — **tokenization**, **part-of-speech
tagging**, and **named entity recognition (NER)**.

This all runs on CPU, in Colab's free tier — no GPU, no local install needed.

## Part 1 — Install & load spaCy's small English model

We use the **small** model (`en_core_web_sm`, ~13MB) on purpose.
It's fast enough to run comfortably even
on a 4GB-RAM laptop, and it's all we need for this course.

```python
import spacy
nlp = spacy.load("en_core_web_sm")
print("spaCy loaded:", spacy.__version__)
```

## Part 2 — Tokenization: breaking a sentence into pieces

A computer doesn't see a sentence the way you do — it sees a list of tokens (roughly: words and
punctuation).

```python
sentence = "I’d like a large pepperoni pizza please"
doc = nlp(sentence)

for token in doc:
    print(token.text)
```

## Part 3 — Part-of-speech (POS) tagging

Now let's see what *kind* of word each token is — noun, verb, proper noun, etc. This is one of the pieces a
chatbot needs to understand what a sentence is actually asking for.

```python
for token in doc:
    print(f"{token.text:12} -> {token.pos_}")
```

## Part 4 — Named Entity Recognition (NER): finding the important "things"

This is the part that matters most for a chatbot: pulling out the specific pieces of information — dates,
places, people, organizations — that it needs to act on.

Try it on a sentence with local names and places.

```python
sentences = [
    "I’d like a large pepperoni pizza please",
    "Send the invoice to Alice Uwimana at the Huye branch by Friday.",
    "Can I get a refund for my order placed in Musanze last month?",
]

for s in sentences:
    doc = nlp(s)
    print(f"\nSentence: {s}")
    if doc.ents:
        for ent in doc.ents:
            print(f"   {ent.text!r:25} -> {ent.label_}")
    else:
        print("   (no entities found)")
```

**Discuss:** Did spaCy correctly recognize the Rwandan place names (Kigali, Huye, Musanze) and the local
name (Alice Uwimana)? Most NLP models are trained mostly on Western/English-language data — this is a real
limitation worth noticing, not a bug you did wrong.

## Part 5 — Start your own intent dataset

This is the beginning of something you'll keep building for the next few weeks: a small labeled dataset of
example sentences, each tagged with the **intent** it represents. In Week 9 you'll use this to train a real
intent classifier.

Pick a domain for your chatbot (e.g. a school helper, a clinic assistant, a shop order-tracker) and write
10–15 example sentences across 3–4 intents, like the example below.

```python
# Example structure — replace with your own domain and sentences
training_data = [
    ("What time does the clinic open?", "opening_hours"),
    ("When are you open on weekends?", "opening_hours"),
    ("I want to book an appointment", "book_appointment"),
    ("Can I see a doctor tomorrow?", "book_appointment"),
    ("How much does a consultation cost?", "pricing"),
    ("What's the price for a check-up?", "pricing"),
]

for text, intent in training_data:
    print(f"{intent:20} | {text}")
```
